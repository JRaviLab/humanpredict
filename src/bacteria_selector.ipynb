```python
def is_bacteria(taxonomy):
    bacteria_ids = {2}  # Initialized as a Set for O(1) lookups
    stopping_points = {1, 131567}

    for key in taxonomy:
        current_node = key
        current_path = []

        while current_node not in stopping_points:
            # If we hit a known bacteria node, everything in our path is bacteria
            if current_node in bacteria_ids:
                bacteria_ids.update(current_path)
                break
            
            current_path.append(current_node)
            
            # Move up the tree safely
            if current_node not in taxonomy:
                break
            current_node = taxonomy[current_node]

    return list(bacteria_ids)


bacteria_ids = is_bacteria(tax_parent_dict)
```
